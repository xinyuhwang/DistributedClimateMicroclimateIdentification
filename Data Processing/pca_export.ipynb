```python
"""
Export full PCA components (all 14) for distributed clustering
"""

import pandas as pd
import pickle
import numpy as np

print("="*80)
print("EXPORTING FULL PCA FEATURES")
print("="*80)

# Load engineered features
print("\nLoading engineered features...")
df_features = pd.read_csv("preprocessed_data/engineered_features_495locs.csv")
print(f"✓ Loaded {len(df_features)} locations")

# Load scaler and PCA model
with open("preprocessed_data/scaler_495locs.pkl", "rb") as f:
    scaler = pickle.load(f)
print("✓ Loaded scaler")

with open("preprocessed_data/pca_model_495locs.pkl", "rb") as f:
    pca = pickle.load(f)
print(f"✓ Loaded PCA model ({pca.n_components_} components)")

# Extract features (exclude lat, lon, cluster)
feature_cols = [col for col in df_features.columns if col not in ['lat', 'lon', 'cluster']]
print(f"\nFound {len(feature_cols)} feature columns")

# Get features and normalize
X = df_features[feature_cols].values
print(f"Feature matrix shape: {X.shape}")

# Apply normalization
X_normalized = scaler.transform(X)
print("✓ Applied normalization")

# Apply PCA transformation
X_pca = pca.transform(X_normalized)
print(f"✓ Applied PCA transformation: {X_normalized.shape} → {X_pca.shape}")

# Create DataFrame with all PCA components
df_pca_full = pd.DataFrame({
    'lat': df_features['lat'],
    'lon': df_features['lon']
})

# Add all PCA components
for i in range(X_pca.shape[1]):
    df_pca_full[f'PC{i+1}'] = X_pca[:, i]

print(f"\n✓ Created DataFrame with {X_pca.shape[1]} PCA components")
print(f"Columns: {df_pca_full.columns.tolist()}")

# Save to CSV
output_file = "preprocessed_data/pca_features_full.csv"
df_pca_full.to_csv(output_file, index=False)
print(f"\n✓ Saved to: {output_file}")

# Display sample
print("\nSample data (first 5 rows):")
print(df_pca_full.head())

print("\n" + "="*80)
print("EXPORT COMPLETE!")
print("="*80)
print(f"\nNow run data preparation again:")
print(f"  The script will use this file with all {X_pca.shape[1]} PCA components")
print("="*80)
```
